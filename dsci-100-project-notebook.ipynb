```r
options(repr.plot.width = 12, repr.plot.height = 8)

march_madness_tidy |>
  ggplot(aes(x = conference)) +
    geom_bar(aes(y = after_stat(count / sum(count))),
             fill = "blue") +
    labs(
        title = "Proportion of Tournament Appearances by Conference",
        x =  "Conference",
        y = "Proportion of Tournament Appearances"
        ) +
    coord_flip() +
    theme_minimal()
```

I plan to use K-nearest neighbors (KNN) regression to predict the number of tournament wins (`tourney_wins`) based on a team's tournament seed (`seed`) and conference win percentage (`conf_wins_pct`). KNN regression is appropriate because the response variable is numerical, and the method predicts outcomes by averaging the values of nearby observations.

One limitation of KNN regression is that its performance depends on the number of neighbors (K). A small K may lead to overfitting, while a large K may lead to underfitting. KNN also requires predictors to be on comparable scales, so I will standardize them before fitting the model. Additionally, KNN regression may produce fractional predictions even though tournament wins are whole numbers.

I will first remove observations with missing values in the variables used for modeling and then split the cleaned data into a 75% training set and a 25% testing set, stratified by `tourney_wins`. I will use 5-fold cross-validation on the training set to evaluate different K values, selecting the one with the lowest average validation root mean squared error (RMSE). Standardization will be performed within the modeling workflow to avoid data leakage.

Finally, I will fit the selected model using the full training set and assess its performance on the testing set using RMSE and mean absolute error (MAE). Lower values indicate better predictions. RMSE penalizes large errors more heavily, while MAE measures the average absolute prediction error.
